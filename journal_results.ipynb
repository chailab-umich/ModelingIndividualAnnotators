# Published journal results

This notebook regenerates Tables II–VIII from the archived CSV outputs. Run all cells from a fresh kernel.

```python
import os
import pandas as pd
import numpy as np

PUBLICATION_GROUP_PATTERN = (
    r'^(?:Mean Activation CCC(?:_ind)?|Mean Valence CCC(?:_ind)?|'
    r'Activation CCC|Valence CCC|Argmax Activation UAR|Argmax Valence UAR|'
    r'Total Variation Distance|Jensen-Shannon Divergence)$'
)

def read_result_csv(path):
    """Read only the aggregate rows needed for the published tables."""
    selected = []
    for chunk in pd.read_csv(
        path,
        usecols=['log_type', 'group', 'scalar_value'],
        chunksize=10_000,
    ):
        keep = chunk['log_type'].str.startswith('Test') & (
            chunk['group'].str.match(PUBLICATION_GROUP_PATTERN)
            | chunk['log_type'].str.contains('loss', case=False)
        )
        selected.append(chunk.loc[keep])
    return pd.concat(selected, ignore_index=True)
```

```python
PUBLICATION_MODELS = ['baseline', 'baseline_aggregate', 'baseline_podcast_1.8', 'baseline_aggregate_podcast_1.8', 'individual_annotator', 'individual_annotator_plus', 'individual_annotator_var', 'individual_annotator_plus_var', 'individual_annotator_100plus', 'individual_annotator_plus_100plus', 'individual_annotator_var_100plus', 'individual_annotator_plus_var_100plus', 'individual_annotator_1000plus', 'individual_annotator_plus_1000plus', 'individual_annotator_var_1000plus', 'individual_annotator_plus_var_1000plus', 'individual_annotator_podcast_1.8', 'individual_annotator_plus_podcast_1.8', 'individual_annotator_var_podcast_1.8', 'individual_annotator_plus_var_podcast_1.8']

RESULTS_ROOT = os.environ.get('MIA_PAPER_RESULTS_ROOT', '.output_files/csv_results')
JOURNAL_RESULTS_DIR = f'{RESULTS_ROOT}/journal_experiments'
```

```python
from datetime import datetime
csvdir = JOURNAL_RESULTS_DIR
missing_models = [model for model in PUBLICATION_MODELS if not os.path.isdir(f'{csvdir}/{model}')]
assert not missing_models, f'Missing publication results: {missing_models}'
models = PUBLICATION_MODELS

def read_csv(model_name):
    p = f'{csvdir}/{model_name}'
    files = os.listdir(p)
    if len(files) > 1:
        print(files)
        datatimes = [datetime.strptime('_'.join(f.split('_')[-4:]), "%d_%m_%Y_%H:%M:%S") for f in files]
        most_recent_file = files[np.argmax(datatimes)]
        # print(files, most_recent_file)
        files = [most_recent_file]
    assert len(files) == 1, f'({model_name}) Expected 1 folder, got {len(files)} ({files})'
    pds = []
    full_path = f'{p}/{files[0]}'
    files = sorted(os.listdir(full_path))
    assert len(files) == 5 and all([f.endswith('.csv') for f in files]), f'Expected 5 files, got {len(files)} ({full_path} | {files})'
    for f in files:
        pds.append(read_result_csv(f'{full_path}/{f}'))
    return pds

results = {
    model: read_csv(model) for model in models
}
```

```python
def get_results_by_log_type(results, log_type):
    return [r[r['log_type'] == log_type] for r in results]

def get_unique_log_types(results):
    log_types = []
    for r in results:
        log_types.append(r['log_type'].unique())
    assert all((l == log_types[0]).all() for l in log_types), f'Expected all log types to be the same, got {log_types}'
    return log_types[0]

results_for_pretty_print = {}
for model in results:
    if model not in results_for_pretty_print:
        results_for_pretty_print[model] = {}
    # print('--------------------------------', model)
    for x in get_unique_log_types(results[model]):
        if 'Training' in x:
            continue
        log_type_results = get_results_by_log_type(results[model], x)
        strat = x.split(' ')[:3][-1]
        dataset_name = x.split(' ')[1]
        # Valid strategies are: aggregate-annotator and similarity, if neither of these are present then it was default/None strategy
        if strat not in ['aggregate-annotator', 'similarity']:
            strat = 'None'
        if dataset_name not in results_for_pretty_print[model]:
            results_for_pretty_print[model][dataset_name] = {}
        if strat not in results_for_pretty_print[model][dataset_name]:
            results_for_pretty_print[model][dataset_name][strat] = {}
        if 'KL-Div' in x:
            continue # Not reporting kl-div loss values for now
        if 'loss' in x:
            # Loss for baseline is not CCC flat
            if model == 'baseline' or model == 'baseline_podcast_1.8':
                continue
            assert all([len(r['group'].unique()) == 1 for r in log_type_results]), f'Expected 1 group, got {len(log_type_results["group"].unique())}'
            act_or_val = 'act' if 'act' in x.lower() else 'val'
            results_for_pretty_print[model][dataset_name][strat][f'{act_or_val} CCC_flat'] = [float(r['scalar_value'].item()) for r in log_type_results]
        else:
            if 'podcast_1.8' not in model:
                results_for_pretty_print[model][dataset_name][strat]['Act CCC_ind'] = [float(r[r['group'].str.contains('Mean Activation CCC')]['scalar_value'].item()) for r in log_type_results]
                results_for_pretty_print[model][dataset_name][strat]['Val CCC_ind'] = [float(r[r['group'].str.contains('Mean Valence CCC')]['scalar_value'].item()) for r in log_type_results]
            results_for_pretty_print[model][dataset_name][strat]['Act CCC'] = [float(r[r['group'].str.match('Activation CCC')]['scalar_value'].item()) for r in log_type_results]
            results_for_pretty_print[model][dataset_name][strat]['Val CCC'] = [float(r[r['group'].str.match('Valence CCC')]['scalar_value'].item()) for r in log_type_results]
```

```python
from scipy.stats import ttest_rel
def get_cell_values(dataset, model, strat, baseline_model, baseline_strat, act_or_val, agg_or_ind):
    result_key = f'{act_or_val} CCC' + ('_ind' if agg_or_ind == 'ind' else '')
    if model not in results_for_pretty_print:
        return 'N/A'
    result = results_for_pretty_print[model][dataset][strat][result_key]

    mean = np.mean(result)
    cell_value = fr'{mean:.3f}$\pm${np.std(result):.3f}'
    if baseline_model is None or baseline_strat is None:
        return cell_value

    baseline_result = results_for_pretty_print[baseline_model][dataset][baseline_strat][result_key]    
    baseline_mean = np.mean(baseline_result)
    pval = ttest_rel(result, baseline_result).pvalue
    significant = pval < 0.05
    if not significant:
        return cell_value

    if mean > baseline_mean:
        return f'{cell_value}*'
    else:
        return fr'{cell_value}$\dagger$'
```
